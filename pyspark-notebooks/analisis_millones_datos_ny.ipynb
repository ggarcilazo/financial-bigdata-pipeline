```python
%sql
SELECT 
    pickup_zip AS zona_recogida_ny,
    COUNT(*) AS total_viajes_reales_procesados,
    AVG(trip_distance) AS distancia_promedio_millas,
    SUM(fare_amount) AS ingresos_totales_tarifa_usd
FROM 
    samples.nyctaxi.trips
GROUP BY 
    pickup_zip
ORDER BY 
    total_viajes_reales_procesados DESC;
```
